```python
def clusterCalculateMLE(cluster1):
    x=np.asanyarray(cluster1)
    n, d = x.shape
    e=sys.float_info.epsilon
    
    mu = x.mean(axis=0)
    sigma= np.cov(x, rowvar=False)
    print(sigma.shape)
    if(n==1):
        sigma=np.zeros((d,d))+ e*np.eye(d)

    L=multivariate_normal.logpdf(x,mean=mu,cov=sigma,allow_singular=True).sum()
    
    print(L)
    return L
    

k = pd.DataFrame()

k=pd.concat([k,Xdf.iloc[1:5,:]])
clusterCalculateMLE(k)


            
```
