```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import numpy as np
```

```python
df=pd.DataFrame()

df['Month']=['Jan','Feb','Mar','Apr','May','Jun']
df['RiderShip(thousands)']=[420,435,460,500,520,545]
df['on-timeRate(%)']=[88,87,86,84,83,82]
df['CustomerComplaints']=[310,325,340,390,420,460]
df['OperatingCost($M)']=[5.2,5.3,5.5,5.8,6.0,6.2]
df
```

```python
plt.bar(df['Month'],df['RiderShip(thousands)'],color='g')

plt.xlabel('Months')
plt.ylabel('RiderShip(k)')
plt.grid(True)
plt.show()

fig, ax = plt.subplots(1,2)
fig.set_figheight(5)
fig.set_figwidth(15)
ax[0].bar(df['Month'],df['on-timeRate(%)'],color='red')
ax[0].set(xlabel='Months')
ax[0].set(ylabel='on-timeRate(%)')
ax[0].grid(True)



ax[1].plot(df['Month'],df['on-timeRate(%)'],color='red')
ax[1].set(xlabel='Months')
ax[1].set(ylabel='on-timeRate(%)')
ax[1].grid(True)
plt.show()
```
